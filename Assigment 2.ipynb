```python
number  = input("Enter any Number : ")
if number.isdigit() or number.startswith('-'):
    number = int(number)
    if number == 0:
        print("Zero")
    elif number > 0:
        print(f"{number} : The number was positive.")
        if number %2==0:
            print("The number was EVEN.")
        else:
            print("The number was ODD.")   
    elif number < 0:
        print(f"{number} : The number was Negative.")
        if number %2==0:
            print("The number was EVEN.")
        else:    
            print("The number was ODD.") 

else:
    print("Invalid Command.")    
```
